```python
data = data.value_counts('Academic Level')
data.plot(kind='pie')
import matplotlib.pyplot as plt
plt.show()
```

```python
data = data.value_counts('Academic Level')

# Create a bar chart
data.plot(kind='bar', figsize=(10, 6))

# Add labels and title for better visualization
import matplotlib.pyplot as plt
plt.xlabel('Academic Level')
plt.ylabel('Count')
plt.title('Count by Academic Level')
plt.xticks(rotation=45)  # Rotate x labels for better readability
plt.grid(axis='y')  # Add horizontal grid lines

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Display the plot
plt.show()
```

```python
data = data.value_counts('Academic Level')

# Create a line chart
# The index will be used as x-axis and the counts as y-axis
data.plot(kind='line', figsize=(10, 6))

# Add labels and title for better visualization
import matplotlib.pyplot as plt
plt.xlabel('Academic Level')
plt.ylabel('Count')
plt.title('Count by Academic Level')
plt.grid(True)  # Add grid for better readability

# Display the plot
plt.show()
```
